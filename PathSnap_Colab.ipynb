# PathSnap training

Use a Google Colab CPU runtime. Upload the project source ZIP and the supplied student data ZIP. The final evaluator partition is reserved for the live demo.

```python
from google.colab import files
from pathlib import Path
from zipfile import ZipFile

uploaded = files.upload()
required = {"pathsnap-source.zip", "Student folder.zip"}
if not required <= uploaded.keys():
    raise ValueError("Upload pathsnap-source.zip and Student folder.zip")
del uploaded
source = Path("/content/pathsnap-source")
with ZipFile("/content/pathsnap-source.zip") as bundle:
    bundle.extractall(source)
%cd /content/pathsnap-source
%pip install -q -r requirements-colab.txt
```

```python
import subprocess
import sys

data = "/content/Student folder.zip"
subprocess.run([sys.executable, "-m", "pathsnap.train", "--data", data], check=True)
```

```python
subprocess.run([sys.executable, "-m", "pathsnap.visualize", "--data", data], check=True)
from IPython.display import Image, display
display(Image(filename="artifacts/model-comparison.png"))
display(Image(filename="artifacts/student-trip-map.png"))
```
